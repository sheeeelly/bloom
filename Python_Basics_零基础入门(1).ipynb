## 📌 单元格 1：变量（Variable）与基本数据类型

**这一步做什么：**
- 创建**变量**来存储数据（类似 Excel 里的单元格或 SPSS 里的变量名）
- 了解三种常见类型：**整数（Integer）**、**浮点数（Float）**、**字符串（String）**

**初学者要点：**
- 用 `=` 给变量赋值，例如 `age = 28`
- 用 `type()` 查看变量类型
- 字符串用引号括起来：单引号 `'...'` 或双引号 `"..."` 都可以

```python
# 数值型变量
age = 28                    # 整数 Integer
grit_score = 3.75           # 浮点数 Float（带小数）

# 字符串 String（文本）
name = "参与者A"
department = "组织行为学"

# 布尔值 Boolean（True / False，表示是/否）
is_leader = True

print("姓名:", name)
print("年龄:", age, "类型:", type(age))
print("Grit 分数:", grit_score, "类型:", type(grit_score))
print("是否领导者:", is_leader, "类型:", type(is_leader))
```

## 📌 单元格 2：简单运算

**这一步做什么：** 对数字做加减乘除，对字符串做拼接。

**初学者要点：** 在组织行为学研究中，您经常需要计算均值、求和等——Python 的运算符与 Excel 类似（`+` `-` `*` `/`）。

```python
# 数值运算
extraversion = 4.2
agreeableness = 3.8
personality_sum = extraversion + agreeableness
personality_mean = personality_sum / 2

print("大五人格两项之和:", personality_sum)
print("大五人格两项均值:", personality_mean)

# 字符串拼接
greeting = "您好，" + name + "！欢迎参加 OB 研究方法工作坊。"
print(greeting)
```

## 📌 单元格 3：列表（List）

**这一步做什么：**
- 用**列表**存储一组有序的数据（类似一列数字）
- 列表用方括号 `[]` 创建，元素之间用逗号分隔

**初学者要点：**
- 索引（Index）从 **0** 开始：`列表[0]` 是第一个元素
- `len(列表)` 返回列表长度（元素个数）

```python
# 5 名参与者的工作满意度评分（1-7 量表）
satisfaction_scores = [5, 6, 4, 7, 5]

print("所有评分:", satisfaction_scores)
print("第一个人的评分:", satisfaction_scores[0])   # 索引从 0 开始
print("最后一个人的评分:", satisfaction_scores[-1])  # -1 表示最后一个
print("参与者人数:", len(satisfaction_scores))

# 用 Python 内置函数计算均值
print("平均满意度:", sum(satisfaction_scores) / len(satisfaction_scores))
```

## 📌 单元格 4：字典（Dictionary）

**这一步做什么：**
- 用**字典**存储「键–值对」（Key–Value Pair），类似一张有两列的小表
- 例如：变量名 → 数值

**初学者要点：** 字典用花括号 `{}`，通过键名取值，如 `字典["键名"]`

```python
# 一位参与者的大五人格简表
participant = {
    "ID": "P001",
    "Extraversion": 4.2,      # 外向性
    "Agreeableness": 3.8,     # 宜人性
    "Conscientiousness": 5.1, # 尽责性
    "Grit": 3.5               # 坚毅
}

print("参与者编号:", participant["ID"])
print("外向性得分:", participant["Extraversion"])
print("全部信息:", participant)
```

## 📌 单元格 5：条件判断（if / elif / else）

**这一步做什么：** 根据条件决定是否执行某段代码。

**初学者要点：**
- `if` 后面跟条件，条件为真（True）时执行
- 注意**缩进（Indentation）**：Python 用缩进（通常 4 个空格）表示代码块，这很重要！

```python
extraversion_score = 5.5

if extraversion_score >= 5:
    print("该参与者外向性偏高（High Extraversion）")
elif extraversion_score >= 3:
    print("该参与者外向性中等（Moderate Extraversion）")
else:
    print("该参与者外向性偏低（Low Extraversion）")
```

## 📌 单元格 6：循环（for Loop）

**这一步做什么：** 用 `for` 循环遍历列表中的每个元素，重复执行相同操作。

**初学者要点：** 循环是批量处理数据的常用方式——例如对 300 份问卷逐一计算。

```python
team_members = ["张三", "李四", "王五", "赵六"]

print("团队成员名单:")
for member in team_members:
    print("  -", member)

# 对数值列表求和
scores = [4, 5, 3, 6, 5]
total = 0
for s in scores:
    total = total + s
print("\n评分总和:", total)
print("评分均值:", total / len(scores))
```

## 📌 单元格 7：函数（Function）

**这一步做什么：**
- 把一段重复使用的代码封装成**函数**
- 使用 `def` 定义函数，用 `return` 返回结果

**初学者要点：** 后续课程会大量使用已有函数（如 `pd.read_csv()`），这里先了解函数的基本概念。

```python
def compute_mean(numbers):
    """计算一组数字的均值（Mean）"""
    return sum(numbers) / len(numbers)

def classify_grit(score):
    """根据 Grit 分数给出简单分类"""
    if score >= 4:
        return "高坚毅 High Grit"
    elif score >= 2:
        return "中等坚毅 Moderate Grit"
    else:
        return "低坚毅 Low Grit"

grit_scores = [3.2, 4.5, 1.8, 3.9, 2.1]
print("Grit 均值:", compute_mean(grit_scores))
print("3.2 的分类:", classify_grit(3.2))
```

## 📌 单元格 10：读取 CSV 文件（read_csv）

**这一步做什么：**
- 用 `pd.read_csv()` 读取课程提供的人格数据文件
- 这是后续 Illustration 2、3 使用的同一数据集

**初学者要点：**
- CSV（Comma-Separated Values）是最常见的数据交换格式

```python
# 读取大五人格 + Grit 数据
df = pd.read_csv("big_five_grit_full_data.csv")

print("数据维度:", df.shape)
print("\n前 5 行:")
df.head()
```

## 📌 单元格 11：选取列与计算描述统计

**这一步做什么：**
- 用 `df["列名"]` 选取某一列
- 用 `df[["列1", "列2"]]` 选取多列
- 计算均值、标准差等描述统计量

**初学者要点：** 在 Illustration 2 中，您将用 **Grit** 作为因变量（Dependent Variable），大五人格作为预测变量（Predictors）。

```python
# 选取大五人格 + Grit 相关列
personality_cols = ["Grit", "Open", "Consc", "Extra", "Agree", "Neuro"]
df_personality = df[personality_cols]

print("=== 描述性统计 describe() ===")
print(df_personality.describe())

print("\n=== Grit 的均值 ===")
print("Grit 均值:", df["Grit"].mean())
print("Grit 标准差:", df["Grit"].std())
```

## 📌 单元格 12：筛选数据（Filtering）

**这一步做什么：** 根据条件筛选满足要求的行，类似 SPSS 中的「选择个案」。

**初学者要点：** 条件表达式放在 `df[条件]` 的方括号内，多个条件用 `&`（且）连接。

```python
# 筛选：外向性 Extra 高于 1 的参与者
high_extra = df[df["Extra"] > 1]
print("外向性 > 1 的人数:", high_extra.shape[0])

# 筛选：Grit 高于均值 且 尽责性 Consc 高于 1
grit_mean = df["Grit"].mean()
selected = df[(df["Grit"] > grit_mean) & (df["Consc"] > 1)]
print("Grit 高于均值且尽责性 > 1 的人数:", selected.shape[0])
```

## 📌 单元格 14：对文本列做简单操作

**这一步做什么：**
- 计算每条回答的**字符数（Character Count）**
- 用 `apply()` 对整列批量应用函数

**初学者要点：** 文本长度是 NLP 分析中最简单的特征之一；Illustration 4 将学习更复杂的文本特征提取方法。

```python
# 新增一列：回答的字符数
df_text["text_length"] = df_text["extraversion_open_response"].apply(len)

print("文本长度描述统计:")
print(df_text["text_length"].describe())

print("\n文本长度与外向性评分的相关性:")
correlation = df_text["text_length"].corr(df_text["extraversion_rating_1_to_7"])
print(f"相关系数 r = {correlation:.3f}")
```
